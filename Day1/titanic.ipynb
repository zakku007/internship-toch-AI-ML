```python
from sklearn.preprocessing import LabelEncoder
```

```python
en=df.copy()
```

```python
label_encoder=LabelEncoder()
en['sex-label']=label_encoder.fit_transform(en['Sex'])
```

```python
one_hot_encoded=pd.get_dummies(en['Embarked'],prefix='Embarked')
en=pd.concat([en,one_hot_encoded],axis=1)
```

```python
pclass_mapping={1:'First',2:'Second',3:'Third'}
en['Pclass_Category'] = en['Pclass'].map(pclass_mapping)
```

```python
print(en[['Sex', 'Sex_Label', 'Embarked', 'Embarked_Q', 'Embarked_S', 'Embarked_C', 'Pclass', 'Pclass_Category']].head())
```
