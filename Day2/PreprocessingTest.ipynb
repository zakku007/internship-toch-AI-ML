```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder,MinMaxScaler,StandardScaler
```

```python
df=pd.read_csv("/content/preprocessing_practice_dataset.csv")
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df['Age_mean_imputed']=df['Age'].fillna(df['Age'].mean())
df['Salary_mean_imputed']=df['Monthly_Salary'].fillna(df['Monthly_Salary'].mean())
df
```

```python
df['Department_Mode']=df['Department'].fillna(df['Department'].mode()[0])
df['City_Mode']=df['City'].fillna(df['Department'].mode()[0])
df
```

```python
df.isnull().sum()
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(10,6))
sns.boxplot(data=df,x="Experience_Years",y="Monthly_Salary",palette='pastel')
plt.title('Boxplot of Monthly salary by Experience')
plt.xlabel('Experience_Years',fontsize=12)
plt.ylabel('Monthly_salary',fontsize=12)
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(10,6))
sns.histplot(data=df,x="Experience_Years",y="Monthly_Salary",palette='pastel')
plt.title('Boxplot of Monthly salary by Experience')
plt.xlabel('Experience_Years',fontsize=12)
plt.ylabel('Monthly_salary',fontsize=12)
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(10,6))
sns.lineplot(data=df,x="Experience_Years",y="Monthly_Salary",palette='pastel')
plt.title('Boxplot of Monthly salary by Experience')
plt.xlabel('Experience_Years',fontsize=12)
plt.ylabel('Monthly_salary',fontsize=12)
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(10,6))
sns.scatterplot(data=df,x="Experience_Years",y="Monthly_Salary",palette='pastel')
plt.title('Boxplot of Monthly salary by Experience')
plt.xlabel('Experience_Years',fontsize=12)
plt.ylabel('Monthly_salary',fontsize=12)
```

```python
sns.set(style='whitegrid')
plt.figure(figsize=(10,6))

gender_counts = df['Gender'].value_counts()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Distribution of Gender')
plt.ylabel('') # Remove y-label for pie chart
```

```python
encoded_data=df.copy()
```

```python
label_encoder=LabelEncoder()
encoded_data['Gender_Label']=label_encoder.fit_transform(encoded_data['Gender'])

encoded_data
```

```python
onehot_encoded=pd.get_dummies(encoded_data['City'],prefix='City_Onehot')
encoded_data=pd.concat([encoded_data,onehot_encoded],axis=1)
encoded_data
```

```python
numerical_columns=['Age','Monthly_Salary']
df.dropna(subset=numerical_columns,inplace=True)

minmax_scaler=MinMaxScaler()
standard_scaler=StandardScaler()

df['Age_MinMax']=minmax_scaler.fit_transform(df[['Age']])
df['Salary_MinMax']=minmax_scaler.fit_transform(df[['Monthly_Salary']])

#apply minmax scaling
df['Age_Standard']=standard_scaler.fit_transform(df[['Age']])
df['Salary_Standard']=standard_scaler.fit_transform(df[['Monthly_Salary']])
df
```
